# Publish aml_gnn_hi_small from a checkpoint dataset

Run as a **transform**: Models panel → model card → *Publish as transform output* → this notebook; input = `aml_model_checkpoints` (if offered) **or** `gnn_training_code` (already panel-registered, also holds the checkpoint); output = `aml_gnn_hi_small` → *Sync and run*. The build publishes with build permissions; the interactive session is denied artifact writes.

```python
from foundry.transforms import Dataset
import os
# Checkpoint lives in BOTH datasets; use whichever one the transform panel accepted as input.
try:
    files = Dataset.get("aml_model_checkpoints").files().download()
    ckpt_path = files["aml_gnn_hi_small.pt"]
except Exception as e:
    print("aml_model_checkpoints not available (", type(e).__name__, ") -> falling back to gnn_training_code")
    files = Dataset.get("gnn_training_code").files().download()
    ckpt_path = files["aml_gnn_hi_small_epoch33.pt"]
print(ckpt_path, os.path.getsize(ckpt_path), "bytes")
```

```python
import train_gnn_foundry as t
from palantir_models.code_workspaces import ModelOutput
adapter, metrics = t.load_checkpoint(ckpt_path)
print(metrics)
model_output = ModelOutput("aml_gnn_hi_small")
model_output.publish(adapter)
```
